```python
import numpy as np

# hidden states
states = ["Walking", "Running", "Resting"]
# observations
observations = ["Low", "Medium", "High"]

# initial state distributions
pi = np.array([0.5, 0.3, 0.2])

# transition probability matrix
A = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
])

# emission probabilities
B = np.array([
    [0.6, 0.3, 0.1],
    [0.1, 0.3, 0.6],
    [0.7, 0.2, 0.1]
])

obs_seq = ["High", "Medium", "Low", "High"]
obs_index = {"Low": 0, "Medium": 1, "High": 2}
O = [obs_index[x] for x in obs_seq]

N = len(states)
T = len(O)

alpha = np.zeros((T, N))

for j in range(N):
    alpha[0][j] = pi[j] * B[j][O[0]]


for t in range(1, T):
    for j in range(N):
        total = 0
        for i in range(N):
            total += alpha[t-1][i] * A[i][j]
        alpha[t][j] = total * B[j][O[t]]

likelihood = np.sum(alpha[T-1])


print("Forward Probability Table(Alpha):\n")
print("Time\tObservation\tWalking\t\t\tRunning\t\t\tResting")
for t in range(T):
    print(f"{t+1}\t{obs_seq[t]}\t\t"
          f"{alpha[t][0]:.6f}\t\t"
          f"{alpha[t][1]:.6f}\t\t"
          f"{alpha[t][2]:.6f}")

print("\nLikelihood:", likelihood)
```
