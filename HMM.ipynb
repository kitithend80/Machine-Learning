```python
#Hidden States
S=input("Enter the hidden states:").split()

#Observation Symbol
V=input("Enter the Observation Symbol:").split()

#Observation Sequence
O=input("Enter the Observation Sequence:").split()

#Initital States
Pi=[float(p) for p in input("Enter the initial states:").split()]

#State Transition Matrix
print()
A=[ [ 0 for _ in range (len(S))] for _ in range (len(S)) ]
for i in range(len(S)):
    for j in range(len(S)):
        A[i][j]=float(input(f"Enter transition probability from state {i+1} to {j+1}:"))

#Printing state transition matrix
print("\nState Transition Matrix:\n")
for i in range(len(S)):
    for j in range (len(S)):
        print(A[i][j],end=" ")
    print()

#Emission Probability Matrix
print()
B=[ [ 0 for _ in range (len(V))] for _ in range (len(S)) ]
for i in range(len(S)):
    for j in range(len(V)):
        B[i][j]=float(input(f"Enter emission probability (P(O={V[j]}|S={S[i]})):"))

#Printing Emission matrix
print("\nEmission Probability Matrix:\n")
for i in range(len(S)):
    for j in range (len(V)):
        print(B[i][j],end=" ")
    print()

#Forward Probability Matrix
print()
alpha=[ [ 0.0 for _ in range (len(O))] for _ in range (len(S)) ]
for i in range(len(S)):
    for j in range(len(O)):
      if j==0:
        alpha[i][j]=Pi[i]*B[i][V.index(O[j])]
      else:
        for k in range(len(S)):
          alpha[i][j]+=alpha[k][j-1]*A[k][i]*B[i][V.index(O[j])]

#Printing Forward Probability
print("\nForward Probability Matrix:\n")
for i in range(len(S)):
    for j in range (len(O)):
        print(f"{alpha[i][j]:.5f}",end=" ")
    print()

#Determing Likelyhood
likelyhood=0
for i in range(len(S)):
  likelyhood+=alpha[i][len(O)-1]

#Printing Likelyhood
print("\nLikelyhood:",likelyhood)
```
