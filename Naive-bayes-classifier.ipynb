```python
import math
import pandas as pd

data = pd.read_csv("iris.csv")
#last column as classlabel
target = data.columns[-1]
#select data for training
train = data.sample(frac=0.8,random_state=1)
#create test data by removing remaining ro
test = data.drop(train.index)
#stores the prior probability
class_prob = {}
total = len(train)

for value in train[target].unique():
    count = len(train[train[target] == value])
    class_prob[value] = count/total

def predict(row):
    probability = {}

    for value in class_prob:
        probability[value] = class_prob[value]
        temp = train[train[target] == value]
        for col in train.columns[:-1]:
            match = len(temp[temp[col] == row[col]])
            probability[value] *= (match+1)/(len(temp)+len(train[col].unique()))
    return max(probability,key=probability.get)

correct = 0
for i,row in test.iterrows():
    if predict(row) == row[target]:
        correct += 1
accuracy = (correct/len(test)) * 100
print("accuracy=",accuracy,"%")

        
                


```
